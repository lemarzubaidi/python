```python
#Target Distribution
sns.countplot(x='target',data = df)

plt.title("Distribution of Heart Disease")
plt.xlabel("Heart Disease")
plt.ylabel("Number of patients")

plt.show()
```

**Target Distibution**

We used the countplot function to determine how many cases of heart disease (1) and how many healthy individuals (0) there were, using the data in the dataframe, and to label the plot and the x- and y-axes.

```python
sns.heatmap(df.corr(numeric_only=True),annot=True)
plt.title("Correlation Matrix")
```

**Correlation Analysis**

We calculate the correlation between the variables and the relationships among all columns, and write the correlation value between each pair of variables in a table.

```python
x=df[["age","sex","cp","trestbps","chol","fbs","restecg","thalach","exang","oldpeak","slope","ca","thal"]]
y=df["target"]
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2)
```

**split**

We entered the features into the x-axis and the target into the y-axis, and performed a subtraction between them, with 20% allocated to the test set and 80% to the training set.

```python
model = LogisticRegression(max_iter=1000)
model.fit(x_train,y_train)
y_pred=model.predict(x_test)
```

**Traning & testing**

Train the model and generate predictions using logistic regression by creating an object named “Model” of type “Logistic Regression,” then training it using the fit function and calculating the prediction for the result after entering test data.

```python
print("Accuracy: ",accuracy_score(y_test,y_pred))
print("Precision: ",precision_score(y_test,y_pred))
print("Recall: ",recall_score(y_test,y_pred))
print("F1-Score: ",f1_score(y_test,y_pred))
print("\nConfusion Matrix:\n: ",confusion_matrix(y_test,y_pred))

cm=confusion_matrix(y_test,y_pred)
sns.heatmap(cm,annot=True)
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.title("confusion Matrix")
```

**Evaluation**

The evaluation measured the quality of the algorithm used and calculated its accuracy, which was 83.9% that means the model was able to correctly identify the condition 83.9% of the time,  percision with a false positive (FP) rate of 79.9% and a recall of 93.6% that means The model detected approximately 93.6% of all patients who actually had heart disease, were detected by the model at a rate of approximately 93.6%. The F1 score was 86.1% that combines precision and recall, The confusion matrix is as follows:
